```python
import numpy as np

arr = np.array([10,20,30,40,50,60])

result = arr[arr > 30]

print(result)
```

```python
arr = np.array([10,20,30,40,50])

print(arr[(arr > 20) & (arr < 50)])
```

```python
names = np.array(["Ram","John","Ravi","Anil"])

print(names[names == "Ram"])
```

```python
arr = np.array([10,20,30,40])

mask = np.array([True,False,True,False])

print(arr[mask])
```

```python
marks = np.array([25,40,65,80,30,55])

passed = marks[marks >= 35]

print(passed)
```
