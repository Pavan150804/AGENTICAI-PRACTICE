```python
import numpy as np
# one d array slicing
arr = np.array([10,20,30,40,50,60])

print(arr)
print(arr[1:4])
print(arr[:4])
print(arr[2:])
print(arr[::2])
print(arr[::-1])
```

```python
#3d slicing
arr = np.arange(24).reshape(2,3,4)

print(arr)
# print(arr[0]) #first block
# print(arr[1]) #second block
# print(arr[0,0]) #first row of first block
# print(arr[:,:,2:]) #last 2 columns of every row
print(arr[:,:2,:]) #First Two Rows of Every Block
print(arr[0,1,1]) #particular element

```
