```python
import torch
a=torch.empty(2,3)
print(a)
print(type(a))
```

```python
#From a List
import torch

a = torch.tensor([1,2,3])

print(a)
print(a.dtype)
```

```python
#All Zeros
a = torch.zeros(2,3)

print(a)

```

```python
# All Ones
a = torch.ones(2,3)

print(a)


```

```python
# Random Tensor
a = torch.rand(2,3)

print(a)

```

```python
# Shape
a = torch.rand(3,4)

print(a.shape)

```

```python
# Reshape
a = torch.arange(12)

print(a)

a = a.reshape(3,4)

print(a)

```

```python
# Tensor Operations

# Addition

a = torch.tensor([1,2,3])
b = torch.tensor([4,5,6])

print(a+b)



# Multiplication

print(a*b)



# Mean

print(a.float().mean())

```
