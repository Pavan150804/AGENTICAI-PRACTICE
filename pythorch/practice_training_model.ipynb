```python

features=df.iloc[:,1:]
target=df["diagnosis"]
```

```python
#standard scaler

std_scaler=StandardScaler()

scaled_df=std_scaler.fit_transform(features)
```

```python
#another way using the map
# map() returns a pandas Series.
# LabelEncoder.fit_transform() returns a NumPy ndarray.
labeled_target=target.map({"M":1,"B":0}).to_numpy()
```

```python
labeled_target
```

```python
#test_train_split data
xtrain,xtest,ytrain,ytest=train_test_split(scaled_df,labeled_target,test_size=0.2)
print(xtrain.shape)
print(xtest.shape)
print(ytest.shape)
print(ytrain.shape)
```

```python
print(xtrain.dtype)
print(type(ytrain))
```

```python
#Numpy arrays to pytorch tensors
xtrain_tensor=torch.from_numpy(xtrain.astype(np.float32))
xtest_tensor=torch.from_numpy(xtest.astype(np.float32))
ytrain_tensor = torch.from_numpy(ytrain.astype(np.float32)).view(-1,1)
ytest_tensor = torch.from_numpy(ytest.astype(np.float32)).view(-1,1)
```

```python
print(xtrain_tensor.shape)
print(xtest_tensor.shape)
print(ytrain_tensor.shape)
print(ytest_tensor.shape)
```

```python
#Defining the model

import torch
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self,features):
        super().__init__()
        self.linear1=nn.Linear(features,5)
        self.relu1=nn.ReLU()
        self.linear2=nn.Linear(5,3)
        self.relu2=nn.ReLU()
        self.linear3=nn.Linear(3,1)
        self.sigmoid1=nn.Sigmoid()
    
    def forward(self,features):
         out = self.linear1(features)
         out = self.relu1(out)
         out = self.linear2(out)
         out = self.relu2(out)
         out = self.linear3(out)
         out = self.sigmoid1(out)
         return out
```

```python
#Training the Network
learning_rate=0.1

#define how many epocha
epochs=100

#define the loss function
loss_function=nn.BCELoss()

#create model
model=MyModel(xtrain_tensor.shape[1])
# print(model.shape)

# define optimizer
optimizer = torch.optim.Adam(model.parameters(),lr=learning_rate)
```

```python
#defining the tensorboard
from torch.utils.tensorboard import SummaryWriter
writer=SummaryWriter(log_dir='runs/cancer')
```

```python
for epoch in range(epochs):
    model.train()
    
    y_pred=model(xtrain_tensor)
    
    loss=loss_function(y_pred,ytrain_tensor)
   
    optimizer.zero_grad()
    
    loss.backward()
    
    optimizer.step()
   
   # train accuracy (no grad needed for the metric)
    with torch.no_grad():
    # (y_pred > 0.5) turns probabilities into 0/1 predictions, shape (455, 1).
    # We compare against y_train_2d which is also (455, 1) -> element-wise match.
       train_acc=((y_pred>0.5).float()==ytrain_tensor).float().mean()

# ---- validation (on the held-out test set) ----
    model.eval()  # put model in evaluation mode
    with torch.no_grad():
      val_pred = model(xtest_tensor)              # shape (114, 1)
      val_loss = loss_function(val_pred, ytest_tensor)
      val_acc = ((val_pred > 0.5).float() == ytest_tensor).float().mean()
      
    writer.add_scalar('Loss/train',loss.item(),epoch+1)
    writer.add_scalar('Loss/val',val_loss.item(),epoch+1)
    writer.add_scalar('Accuracy/train', train_acc.item(), epoch + 1)
    writer.add_scalar('Accuracy/val', val_acc.item(), epoch + 1)
    
  # print metrics in each epoch
    print(f'Epoch: {epoch + 1}, '
    f'Train Loss: {loss.item():.4f}, Val Loss: {val_loss.item():.4f}, '
    f'Train Acc: {train_acc.item():.4f}, Val Acc: {val_acc.item():.4f}')

writer.close()
```
