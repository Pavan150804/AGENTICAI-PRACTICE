```python
import pandas as pd
import numpy as np

# Create Series
series = pd.Series([10,20,30,40,50],
                   index=['A','B','C','D','E'])

print("Original Series")
print(series)

# Access
print(series["B"])

# Slice
print(series["B":"D"])

# iloc
print(series.iloc[0:3])

# Filter
print(series[series>30])

# Insert
series["F"]=60

# Delete
series=series.drop("C")

print(series)

# Size
print(series.size)

# Unique
print(series.nunique())
```

```python

data_with_nan= {
    "Name": ["Alice","Bob","Charlie"],
    "Age": [20,np.nan,22],
    "Marks": [85,90,np.nan]
}

df=pd.DataFrame(data_with_nan)
print(df)
```

```python
#handling missing values
df["Age"]=df["Age"].fillna(df["Age"].mean())
```

```python
print(df)
```

```python
df["Marks"]=df["Marks"].fillna(99)
```

```python
sd=pd.read_excel("Book1.xlsx")
print(sd)
```

```python
# print(sd[:3])#First 3 Rows
# print(sd[2:6]) #Rows 2 to 5
# print(sd[-3:])#last 3 rows
# print(sd["Name"]) #select one column
print(sd[["Name","Marks"]]) #select multiple columns
```

```python
# using iloc

# print(sd.iloc[0:5])#First 5 rows
# print(sd.iloc[0:5,0:3]) #First 5 rows and first 3 columns
# print(sd.iloc[:,0:2])#All rows, first 2 columns
print(sd.iloc[4,1]) #single value
```

```python
#using loc
# print(sd.loc[2:6]) #Rows 2 to 6
# print(sd.loc[:,["Name","Marks"]])#Select Name and Marks
print(sd.loc[2:5,["Name","Department"]])
```

```python
#filtering
print(sd[sd["Marks"]>80]["Name"])
print(sd[(sd["Department"]=="CSE") & (sd["Marks"]>80)]) #CSE Students with Marks > 80
```

```python
# print(sd.iloc[::2]) #Every Second Row
print(sd.iloc[::-1]) #Reverse the DataFrame
```

```python
#groupby("colmun_name")

print(sd.groupby("Department")["Marks"].mean())
```

```python
print(sd.groupby("Department")["Marks"].agg(["mean","max","min","count"]))
```
